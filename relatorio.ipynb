# Aurora Sieger - Atividade 1

A primeira atividade do PBL consiste em gerar dados de telemetria e validar, utilizando Python, o dataset gerado, resultando na tomada de decisão entre **decolar** ou **abortar decolagem**.

## Geração de Dados

Para gerar os dados de telemetria, precisamos rodar o arquivo `data_generator.py`, afim de termos um dataset gerado com dados randomicos dentro da faixa de valores pré definidos.

O script `data_generator.py` possui uma flag opcional `-l` ou `--lines`, podendo passar a quantidade de linhas que queremos gerar para o dataset.

A geração dos dados é feita de forma simples, utilizando valores gerados aleatórios através da biblioteca `random`, presente de forma nativa no ecossistema Python e escrevendo em um arquivo CSV.

Os campos de telemetria e seus valores mínimos e máximos considerados em questão são:

- Temperatura interna (entre 18 e 26 graus)
- Temperatura externa (entre -100 e 100 graus)
- Integridade estrutural (apenas 1 como valor aceitável)
- Nível de energia (entre 50% e 100%)
- Pressão dos tanques (entre 100 e 140 psi)
- Status dos módulos críticos (apenas 1 como valor aceitável)

Ao final do script, é gerado o arquivo `data.csv` na pasta do projeto. Esse arquivo é o nosso dataset gerado.

## Algoritmo de validação

O algoritmo de validação presente no arquivo principal entitulado `main.py` apresenta um fluxo simples, onde o arquivo CSV passa por um processamento e verificação, assim como o cálculo da média de todos os campos considerados para então decidir positivamente ou negativamente para a decolagem da nave.

### Fluxo do algoritmo

O fluxo começa lendo o arquivo `data.csv` gerado pelo script `data_generator.py` mencionado anteriormente, passando para a função `validate_data`, que tem duas funções internas: separar os headers da planilha e gerar as médias com os valores passados.
